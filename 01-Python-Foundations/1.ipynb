```python
#TUPLES SETS
'''8. Employee skills

You have:

employee_1 = {"Python", "SQL", "Git"}
employee_2 = {"Python", "Java", "Git"}

Find:

skills both employees have
skills only employee 1 has
all unique skills

This is a very practical use of sets.'''


employee_1 = {"Python", "SQL", "Git"}
employee_2 = {"Python", "Java", "Git"}

print(employee_1 & employee_2)
print(employee_1.difference(employee_2))
print(employee_1.union(employee_2))
```

```python
#10. Simple expense summary

expenses = {
    "food": 3500,
    "rent": 12000,
    "travel": 1500,
    "shopping": 2500
}

# Total
total = sum(expenses.values())
print("Total:", total)

# Highest expense category
highest = max(expenses, key=expenses.get)
print("Highest:", highest)

# Lowest expense category
lowest = min(expenses, key=expenses.get)
print("Lowest:", lowest)

# Average
average = total / len(expenses)
print("Average:", average)
```
